# EPISENCE: Disease Risk Prediction Model
This notebook demonstrates the training, optimization, and evaluation of the epidemiological risk prediction model for the Pune region.

### Objective:
Predict the `composite_risk_score` for a specific ward (zone) 7 days into the future to enable proactive public health response.

### Key Improvements for Maximum Accuracy:
- **Hyperparameter Tuning**: Using Grid Search to find the optimal learning rate and tree depth.
- **TimeSeriesSplit**: Prevents temporal data leakage, ensuring the model generalizes to future outbreaks.
- **Feature Engineering**: Integration of weather lag effects and rolling epidemiological trends.
- **Huber Loss**: Robustness against extreme outliers (sudden spikes in cases).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import json
import joblib
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error

%matplotlib inline
sns.set_theme(style="darkgrid", palette="muted")
```

## 1. Data Loading & Preprocessing

```python
DATA_DIR = "EPISENCE\backend\data"

def load_json(filename):
    with open(os.path.join(DATA_DIR, filename), 'r') as f:
        return json.load(f)

zones = pd.DataFrame(load_json("zones.json"))
weather = pd.DataFrame(load_json("weather_data.json"))
cases = pd.DataFrame(load_json("hospital_cases.json"))
risk = pd.DataFrame(load_json("risk_scores.json"))

# Convert dates
weather['date'] = pd.to_datetime(weather['date'])
cases['date'] = pd.to_datetime(cases['date'])
risk['date'] = pd.to_datetime(risk['date'])

# Aggregate cases by zone and date
cases_agg = cases.groupby(['zone_id', 'date'])['case_count'].sum().reset_index()

# Merge datasets
df = pd.merge(weather, cases_agg, on=['zone_id', 'date'], how='left')
df['case_count'] = df['case_count'].fillna(0)

# Merge with target risk score
df = pd.merge(df, risk[['zone_id', 'date', 'composite_score']], on=['zone_id', 'date'], how='left')
df['composite_score'] = df['composite_score'].ffill().bfill()

print(f"Loaded {len(df)} initial samples.")
```

## 2. Advanced Feature Engineering
We create temporal features that capture the 'momentum' of a disease outbreak.

```python
df = df.sort_values(['zone_id', 'date'])
new_df_list = []

for zone_id, group in df.groupby('zone_id'):
    group = group.copy()
    group['month'] = group['date'].dt.month
    group['is_monsoon'] = group['month'].isin([6, 7, 8, 9]).astype(int)
    
    # Rolling averages for weather
    group['temp_rolling_7d'] = group['temperature_avg'].rolling(window=7).mean()
    group['rain_rolling_14d'] = group['rainfall_mm'].rolling(window=14).sum()
    
    # Exponentially weighted moving average for cases (detects trends faster)
    group['cases_ewma_7d'] = group['case_count'].ewm(span=7).mean()
    
    # Interaction feature: Heat + Humidity (ideal for vector breeding)
    group['vector_breeding_index'] = group['temperature_avg'] * group['humidity_avg'] / 100
    
    # Lagged targets (Predicting 7 days ahead)
    group['target_risk_7d'] = group['composite_score'].shift(-7)
    
    new_df_list.append(group)

df = pd.concat(new_df_list).dropna(subset=['target_risk_7d']).fillna(0)
print(f"Feature engineering complete. Total samples: {len(df)}")
```

## 3. Hyperparameter Tuning
We use `GridSearchCV` with `TimeSeriesSplit` to find the best model parameters.

```python
features = [
    'temperature_avg', 'humidity_avg', 'rainfall_mm', 'case_count',
    'month', 'is_monsoon', 'temp_rolling_7d', 'rain_rolling_14d', 
    'cases_ewma_7d', 'vector_breeding_index'
]

X = df[features]
y = df['target_risk_7d']

tscv = TimeSeriesSplit(n_splits=3)
param_grid = {
    'n_estimators': [100, 200],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 4, 5],
    'loss': ['huber']
}

grid_search = GridSearchCV(
    GradientBoostingRegressor(random_state=42),
    param_grid, 
    cv=tscv, 
    scoring='neg_mean_absolute_error',
    n_jobs=-1
)

print("Tuning hyperparameters...")
grid_search.fit(X, y)

best_model = grid_search.best_estimator_
print(f"Best Parameters: {grid_search.best_params_}")
```

## 4. Evaluation & Visualization
How do we know the prediction is correct?

```python
y_pred = best_model.predict(X)
mae = mean_absolute_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"Overall MAE: {mae:.4f}")
print(f"Overall R2 Score: {r2:.4f}")

# Visualization 1: Actual vs Predicted
plt.figure(figsize=(12, 6))
plt.scatter(y, y_pred, alpha=0.5, color='coral')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'k--', lw=2)
plt.xlabel('Actual Risk Score')
plt.ylabel('Predicted Risk Score')
plt.title('Prediction Accuracy: Actual vs Predicted')
plt.show()

# Visualization 2: Feature Importance
plt.figure(figsize=(10, 6))
importances = pd.Series(best_model.feature_importances_, index=features)
importances.sort_values().plot(kind='barh', color='teal')
plt.title('Key Drivers of Disease Risk')
plt.show()
```

## 5. Model Export
Exporting the model for use in the EPISENCE Dashboard.

```python
MODEL_PATH = "../models/risk_predictor.joblib"
joblib.dump(best_model, MODEL_PATH)
print(f"Optimized model saved to {MODEL_PATH}")
```
